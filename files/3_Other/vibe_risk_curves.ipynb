# Enterprise Vibe Coding

Business and IT leaders must make a risk decision: does the risk of vibe coding outweigh the risk of not vibe coding?

## Vibe Coding

The risks of vibe coding are:
1) **Quality**: Will the quality of the resulting software be immediately worse (buggy)? [1](https://www.databricks.com/blog/passing-security-vibe-check-dangers-vibe-coding)
2) **Security**: Does using an LLM in the SDLC introduce new risks to my enterprise? [2](https://www.wiz.io/blog/critical-vulnerability-base44) [3](https://www.catonetworks.com/blog/curxecute-rce/) [4](https://i.blackhat.com/BH-USA-25/Presentations/US-25-Lynch-From-Prompts-to-Pwns.pdf)
3) **Workforce Management**: Am I shooting myself in the foot by reducing talent development and training opportunities? [5](https://www.finalroundai.com/blog/ai-vibe-coding-destroying-junior-developers-careers)

Ack that "vibe coding" is itself a spectrum and there are ways to do it with more or less rigor.

## Not Vibe Coding
1) Missing a generational efficiency opportunity and being scorned as a luddite while your company falls behind.

Encourage these leaders to plot themselves on the following graph.

```python
import matplotlib.pyplot as plt

x = [0, 1]
y = [0, 1]

fig, ax = plt.subplots()

ax.set_xlabel("Desire to Embrace Vibe Coding")
ax.set_ylabel("Risk Tolerance")
ax.set_xticks([])
ax.set_yticks([])

for spine in ["top", "right"]:
    ax.spines[spine].set_visible(False)

ax.spines["left"].set_position(('data', 0))
ax.spines["bottom"].set_position(('data', 0))
ax.set_aspect('equal', adjustable='box')

ax.plot(x, y)
plt.tight_layout()
plt.show()
```

They're going to put themselves somewhere on it. It's unitless, there's no pressure, but that should be a good vibe check. Some folks might use their risk tolerance to inform their appetite for vibe coding, others may do the opposite, but now you can explore how every enterprise has existing characteristics that change the shape of that line.

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(0, 1, 400)
y_up = 1 - np.sqrt(1 - x**2)
y_down = np.sqrt(1 - (1 - x)**2)

fig, axes = plt.subplots(1, 2, figsize=(10, 4))

titles = ["Mature Enterprise Culture and Controls", "Not"]
functions = [y_up, y_down]

x_mark = 0.75

for ax, y, title in zip(axes, functions, titles):

    # Labels
    ax.set_xlabel("Vibe Coding Adoption")
    ax.set_ylabel("Risk")

    ax.set_xticks([])
    ax.set_yticks([])

    for spine in ["top", "right"]:
        ax.spines[spine].set_visible(False)

    ax.spines["left"].set_position(('data', 0))
    ax.spines["bottom"].set_position(('data', 0))
    ax.set_aspect('equal', adjustable='box')
    ax.set_xlim(0, 1)
    ax.set_ylim(0, 1)

    ax.plot(x, y)

    ax.axvline(x_mark, color='red', linewidth=2)
    y_mark = np.interp(x_mark, x, y)
    ax.hlines(y_mark, xmin=0, xmax=x_mark, colors='red', linestyles='dotted')
    ax.plot(x_mark, y_mark, 'ro')

    ax.set_title(title)

plt.tight_layout()
plt.show()
```

Ultimately your enterprise's ability to safely adopt vibe coding is limited by the maturity of your culture and controls -- those same fundamentals that we're essential before AI.

**Do you have documented coding and testing standards?** Those same things you needed to ensure code quality as developers turned over are incredibly useful for steering coding agents.

**Do you have mature sandboxing strategies and dev/stage/prod isolation?** These abstractions will allow you to unleash automatic execution that's essential for getting the best performance out of agents.

**Do you have an automation strategy for credential and identity management, host provisioning, CI/CD, etc?** These things allow you to horizontally scale agents and appropriately hook them into enabling systems.

These things all shift your curve to look like the one on the left, enabling your organization to lean into AI more with minimal marginal risk. Sure, you _can_ adopt vibe coding without those things, but then you look like the curve on the right where each new unit of AI moves you further up the risk axis.

In fact, I'd go so far as to say that it's leaders' main job to shape this curve. Sure, we may look to them for risk decisions (plotting the line), but their real responsibility is shaping the strategy that forms the curve and positions organizations to adapt to things like AI-enabled development.
